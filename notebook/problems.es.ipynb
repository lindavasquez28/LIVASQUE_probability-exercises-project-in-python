# Ejercicios de probabilidad

## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
# Solución
import numpy as np

def simular_lanzamiento_dados(num_simulaciones=1000, 
                              caras_dado=6, 
                              valor_objetivo=7):

    dado1 = np.random.randint(1, caras_dado + 1, num_simulaciones)
    dado2 = np.random.randint(1, caras_dado + 1, num_simulaciones)
    
    sumas = dado1 + dado2

    casos_exitosos = np.sum((sumas % 2 == 0) | (sumas > valor_objetivo))
    
    # Calculamos la probabilidad
    probabilidad = casos_exitosos / num_simulaciones
    return probabilidad

# Ejecución
prob_dados = simular_lanzamiento_dados(num_simulaciones=1000)
print(f"La probabilidad estimada (par o mayor a 7) es: {np.round(prob_dados * 100, 2)}%")
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Toas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
# Parámetros iniciales
bolas_color = {
    "Blanca": 10,
    "Roja": 20,
    "Verde": 30
}

# Dinámica de la caja
caja_bolas = {}
indice = 0
for color, cantidad in bolas_color.items():
    for _ in range(cantidad):
        caja_bolas[indice] = color
        indice += 1

print(f"Caja creada con {len(caja_bolas)} bolas.")

# Función de simulación
def simulacion(caja, num_simulaciones=1000, num_extracciones=5):
    caso_3_blancas_2_rojas = 0
    caso_mismo_color = 0
    total_bolas = len(caja)
    
    for _ in range(num_simulaciones):
        indices_aleatorios = np.random.randint(0, total_bolas, num_extracciones)
       
        colores_extraidos = [caja[i] for i in indices_aleatorios]
        
        blancas = colores_extraidos.count("Blanca")
        rojas = colores_extraidos.count("Roja")
        verdes = colores_extraidos.count("Verde")
        
        # Condición: 3 blancas y 2 rojas
        if blancas == 3 and rojas == 2:
            caso_3_blancas_2_rojas += 1
            
        # Condición: Todas del mismo color
        if blancas == num_extracciones or rojas == num_extracciones or verdes == num_extracciones:
            caso_mismo_color += 1

    # Cálculo de probabilidades finales
    prob_1 = caso_3_blancas_2_rojas / num_simulaciones
    prob_2 = caso_mismo_color / num_simulaciones
    
    return prob_1, prob_2

# Ejecución
prob_3b2r, prob_iguales = simulacion(caja_bolas, num_simulaciones=1000, num_extracciones=5)

print(f"La probabilidad de sacar 3 blancas y 2 rojas es: {np.round(prob_3b2r * 100, 2)}%")
print(f"La probabilidad de que todas sean del mismo color es: {np.round(prob_iguales * 100, 2)}%")
```
